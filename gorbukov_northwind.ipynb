## Создание массива и его свойства

NumPy хранит числовые данные в массивах. После `import numpy as np` функции библиотеки доступны через `np`.\
`np.array(...)` создаёт массив из списка. `type(...)` показывает тип объекта, `dtype` — тип элементов. `size` — число элементов, `ndim` — число измерений, `shape` — форма. Свойства записываются без круглых скобок; методы, например `sum()`, вызываются со скобками.

### Задание 1.01 Первый массив

```python
# Задание 1.01 Первый массив
# Ваш код
import numpy as np

a = np.array(numbers)
print(a)
```

### Задание 1.03 Размер и форма

Выведите `a.size,` `a.ndim`, `a.shape`. Объясните, почему у формы одномерного массива один компонент.


```python
# Задание 1.03 Размер и форма
# Ваш код
print("a.size:", a.size)
print("a.ndim:", a.ndim)
print("a.shape:", a.shape)
```

>**Ваш ответ:**
>
>У формы одномерного массива (`shape`) один компонент, потому что для описания его размеров достаточно одного числа — количества элементов в нем (в нашем случае `6`). В отличие от двумерной таблицы (где нужны строки и столбцы) или трехмерного куба, здесь есть только одно направление (длина), поэтому форма записывается как кортеж из одного элемента: `(6,)`.

## Массивы по заданному правилу

- `np.zeros(n)`, `np.ones(n)` и `np.full(n, value)` создают массивы длины `n`. 
- `np.arange(start, stop, step)` задаёт шаг и исключает правую границу `stop`. 
- `np.linspace(start, stop, num)` задаёт число точек и по умолчанию включает обе границы.

### Задание 1.04 Нули единицы и одинаковые значения

Создайте и выведите три массива длины 5: из нулей, из единиц и из числа 8.

```python
# Задание 1.04 Нули единицы и одинаковые значения
# Ваш код
zeros_arr = np.zeros(5)
ones_arr = np.ones(5)
eights_arr = np.full(5, 8)

print(zeros_arr)
print(ones_arr)
print(eights_arr)
```

### Задание 1.05 Последовательности с заданным шагом

Через `np.arange` получите числа от 0 до 9 включительно и чётные числа от 2 до 20 включительно. Объясните выбор правой границы.

```python
# Задание 1.05 Последовательности с заданным шагом
# Ваш код
arr_0_to_9 = np.arange(0, 10)
arr_even = np.arange(2, 22, 2)

print(arr_0_to_9)
print(arr_even)
```

>**Ваш ответ:**
>
>Правая граница (`stop`) выбирается на единицу больше последнего нужного числа (или на один шаг больше). Это связано с тем, что функция `np.arange` н**е включает** правую границу в результат. Чтобы получить числа до 9 включительно, мы указываем `stop=10`. Чтобы получить чётные до 20 включительно, указываем `stop=22` при шаге 2.

### Задание 1.06 Равномерные точки

Через `np.linspace` получите ровно 5 чисел от 0 до 1, включая обе границы.

```python
# Задание 1.06 Равномерные точки
# Ваш код
arr_linspace = np.linspace(0, 1, 5)

print(arr_linspace)
```

> **Ваш ответ:**
> 
> Функция `np.linspace(0, 1, 5)` возвращает 5 равномерно распределенных точек. Поскольку по умолчанию `endpoint=True`, обе границы (0 и 1) включаются в результат, а шаг между точками вычисляется автоматически: `(1 - 0) / (5 - 1) = 0.25`.

## Индексы и срезы

Индексы начинаются с нуля. Индекс `-1` означает последний элемент. Срез `start:stop:step` выбирает элементы от `start` до `stop`, не включая `stop`. Пропущенная граница означает начало или конец массива. Отрицательный шаг позволяет идти в обратном порядке.

### Задание 1.07 Отдельные элементы

Получите первый, третий и последний элементы `a` с помощью индексов.

```python
# Задание 1.07 Отдельные элементы
# Ваш код
print(a)
print("Первый элемент (индекс 0):", a[0])
print("Третий элемент (индекс 2):", a[2])
print("Последний элемент (индекс -1):", a[-1])
```

### Задание 1.08 Непрерывные фрагменты

Получите первые три элемента, последние три элемента и элементы со второго по четвёртый включительно.

```python
# Задание 1.08 Непрерывные фрагменты
# Ваш код
print(a)
print("Первые три:", a[:3])
print("Последние три:", a[-3:])
print("Со второго по четвертый включительно:", a[1:4])
```

### Задание 1.09 Шаг и обратный порядок

Получите каждый второй элемент, начиная с первого. Затем выведите все элементы в обратном порядке.

```python
# Задание 1.09 Шаг и обратный порядок
# Ваш код
print(a)
print("Каждый второй, начиная с первого:", a[::2])
print("В обратном порядке:", a[::-1])
```

## Изменение копии массива

Присваивание меняет элемент или выбранный фрагмент. Для независимого массива используйте `copy()`. Запись `other = a` связывает два имени с одним объектом.

### Задание 1.10 Замена элементов

Создайте `changed = a.copy()`. Замените в `changed` второй элемент на 100, затем элементы с третьего по пятый на −1. Выведите `a` и `changed`; исходный массив должен сохранить свои значения.

```python
# Задание 1.10 Замена элементов
# Ваш код
changed = a.copy()

# Второй элемент (порядковый номер 2) имеет индекс 1
changed[1] = 100

# Элементы с третьего по пятый включительно (порядковые номера 3, 4, 5) 
# имеют индексы 2, 3, 4. В срез [2:5] правый индекс 5 не включается, 
# поэтому будут изменены именно индексы 2, 3 и 4.
changed[2:5] = -1

print("Исходный массив a:", a)
print("Измененный массив changed:", changed)
```
