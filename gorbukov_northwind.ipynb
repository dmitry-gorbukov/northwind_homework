## Изменение формы и вычисления по осям

`reshape(rows, columns)` меняет форму, сохраняя число элементов. `reshape(-1)` делает массив одномерным.

В матрице `axis=0` означает свёртку строк: останется по одному значению для каждого столбца. `axis=1` означает свёртку столбцов: останется по одному значению для каждой строки.

### Задание 1.19 Изменение формы

Создайте через `np.arange` числа от 1 до 12 включительно. Преобразуйте их в матрицу 3 × 4, затем снова в одномерный массив. Объясните, почему матрица 5 × 3 из этих чисел не получится.

````python
# Задание 1.19 Изменение формы
# Мой код

arr_1_12 = np.arange(1, 13)
print("Исходный одномерный массив:")
print(arr_1_12)

matrix_3x4 = arr_1_12.reshape(3, 4)
print("\nМатрица 3x4:")
print(matrix_3x4)

arr_flat = matrix_3x4.reshape(-1)
print("\nСнова одномерный массив:")
print(arr_flat)
````

> **Мой ответ:**
> 
> Матрица 5 × 3 не получится, потому что в ней должно быть 15 элементов (5 * 3 = 15), а в исходном массиве от 1 до 12 включительно всего 12 элементов. Метод `reshape` может только менять форму, но не может создавать или удалять данные, поэтому общее количество элементов должно совпадать.

### ★ 1.S1 Срез и независимая копия

Создайте `x = np.arange(6)`, затем `part = x[1:4]` и `independent = x[1:4].copy()`. Выполните `part[0] = -10` и `independent[-1] = 99`. До запуска предскажите значения всех трёх массивов после изменений. Объясните, почему изменение `part` влияет на `x`.

**Предсказание результатов:**

> **Мой ответ:**
> 
> * `x` станет: `[ 0, -10,   2,   3,   4,   5]`
> * `part` станет: `[-10,   2,   3]`
> * `independent` станет: `[ 1,  2, 99]`

````python
# ★ 1.S1 Срез и независимая копия
# Мой код

x = np.arange(6)
part = x[1:4]
independent = x[1:4].copy()

part[0] = -10
independent[-1] = 99

print("x:", x)
print("part:", part)
print("independent:", independent)
````

> **Почему изменение `part` влияет на `x`?**
> 
> В NumPy базовая срезка (срез) возвращает **представление (view)**, а не копию данных. Переменная `part` — это просто «окно», которое смотрит на ту же самую область памяти, что и исходный массив `x`. Поэтому, когда мы меняем `part[0]`, мы на самом деле меняем элемент в памяти, на который ссылается и `x`. 
> Чтобы получить независимый массив, мы явно вызвали метод `.copy()`, который физически скопировал данные в новую область памяти — поэтому изменения в `independent` не затронули `x`.

````python
# ★ 1.S2 Шахматная доска
# Мой код

# 1. Создаем матрицу 8x8 из нулей
board = np.zeros((8, 8), dtype=int)

# 2. Присваиваем единицы "в шахматном порядке" через срезы с шагом
# В четных строках (0, 2, 4, 6) единицы стоят на нечетных индексах (1, 3, 5, 7)
board[::2, 1::2] = 1
# В нечетных строках (1, 3, 5, 7) единицы стоят на четных индексах (0, 2, 4, 6)
board[1::2, ::2] = 1

print(board)
````

### ★ 1.S3 Таблица умножения

Создайте таблицу умножения чисел от 1 до 10 без циклов. Подсказка: умножение массивов форм `(10, 1)` и `(1, 10)` даёт результат формы `(10, 10)`. Этот механизм называется broadcasting. Объясните, какие значения перемножаются для элемента на пересечении третьей строки и седьмого столбца.

## Series и DataFrame

`Series` — одномерный набор значений с метками индекса. `DataFrame` — таблица из столбцов. В словаре для создания таблицы ключи задают названия столбцов, а списки — их значения. `loc` выбирает по меткам, `iloc` — по позициям, которые начинаются с нуля. ### Задание 2.04 Строка по метке и по позиции

Из `mini` получите строку с меткой 103 и строку с позицией 1. Затем получите строки через `mini.loc[101:103]` и `mini.iloc[0:2]`. Объясните, почему границы этих срезов записаны по-разному.

````python
# Задание 2.04 Строка по метке и по позиции
# Мой код

print("Строка с меткой 103 (loc):")
print(mini.loc[103])

print("\nСтрока с позицией 1 (iloc):")
print(mini.iloc[1])

print("\nСрез mini.loc[101:103]:")
print(mini.loc[101:103])

print("\nСрез mini.iloc[0:2]:")
print(mini.iloc[0:2])
````

> **Мой ответ:**
> 
> Границы срезов записаны по-разному, потому что `loc` работает с **метками индекса**, а `iloc` — с **позициями** (номерами строк, начиная с нуля). У `loc` правая граница **включается** в результат: `mini.loc[101:103]` вернёт строки с метками 101 и 103. У `iloc` правая граница **исключается**, как в обычных срезах Python: `mini.iloc[0:2]` вернёт позиции 0 и 1 (то есть те же строки 101 и 103). Таким образом, оба среза дают одинаковый результат, но по разным правилам. Если бы индекс был не `[101, 103, 107]`, а, например, `[10, 20, 30]`, то `loc[10:20]` включил бы обе строки, а `iloc[0:1]` вернул бы только первую — тогда результаты различались бы. Различие в поведении правой границы — главное, что нужно запомнить: `loc` — «включая конец», `iloc` — «не включая конец».

````python
# Задание 2.16 Несколько допустимых значений
# Мой код

# Способ 1: через логическое ИЛИ (|)
cat_1_or_2_pipe = products[(products['categoryID'] == 1) | (products['categoryID'] == 2)]

# Способ 2: через isin()
cat_1_or_2_isin = products[products['categoryID'].isin([1, 2])]

# Вывод списков ID
print("Вывод списков ID через |:", cat_1_or_2_pipe['productID'].tolist())
print("Вывод списков ID через isin:", cat_1_or_2_isin['productID'].tolist())

# Проверка на равенство
is_equal = cat_1_or_2_pipe['productID'].equals(cat_1_or_2_isin['productID'])
print("\nНаборы productID совпадают:", is_equal)
````

Создайте `adjusted = products.copy()`. Только в `adjusted` увеличьте цены действующих товаров категории 1 на 10% через `loc`. Пересчитайте `stock_value` этой копии. Сравните старую и новую цену товара с `productID = 1`.

````python
# Задание 2.18 Изменение цены в отдельной копии
# Мой код

# 1. Создаем независимую копию
adjusted = products.copy()

# 2. Создаем маску для строк, которые нужно изменить
mask = (adjusted['discontinued'] == 0) & (adjusted['categoryID'] == 1)

# 3. Меняем цены через loc (умножаем на 1.10)
adjusted.loc[mask, 'unitPrice'] *= 1.10

# 4. Пересчитываем stock_value в измененной копии
adjusted['stock_value'] = adjusted['unitPrice'] * adjusted['unitsInStock']

# 5. Сравниваем старую и новую цену для товара с productID = 1
old_price = products.loc[products['productID'] == 1, 'unitPrice'].values[0]
new_price = adjusted.loc[adjusted['productID'] == 1, 'unitPrice'].values[0]

print(f"Старая цена товара ID 1: {old_price:.2f}")
print(f"Новая цена товара ID 1: {new_price:.2f}")
print(f"Разница: {new_price - old_price:.2f}")
````
