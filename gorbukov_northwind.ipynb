## Изменение формы и вычисления по осям

`reshape(rows, columns)` меняет форму, сохраняя число элементов. `reshape(-1)` делает массив одномерным.

В матрице `axis=0` означает свёртку строк: останется по одному значению для каждого столбца. `axis=1` означает свёртку столбцов: останется по одному значению для каждой строки.

### Задание 1.19 Изменение формы

````python
# Задание 1.19 Изменение формы
# Мой код

arr_1_12 = np.arange(1, 13)
print("Исходный одномерный массив:")
print(arr_1_12)

matrix_3x4 = arr_1_12.reshape(3, 4)
print("\nМатрица 3x4:")
print(matrix_3x4)

arr_flat = matrix_3x4.reshape(-1)
print("\nСнова одномерный массив:")
print(arr_flat)
````

### ★ 1.S1 Срез и независимая копия

Создайте `x = np.arange(6)`, затем `part = x[1:4]` и `independent = x[1:4].copy()`. Выполните `part[0] = -10` и `independent[-1] = 99`. До запуска предскажите значения всех трёх массивов после изменений. Объясните, почему изменение `part` влияет на `x`.

**Предсказание результатов:**

> **Мой ответ:**
> 
> * `x` станет: `[ 0, -10,   2,   3,   4,   5]`
> * `part` станет: `[-10,   2,   3]`
> * `independent` станет: `[ 1,  2, 99]`

````python
# ★ 1.S1 Срез и независимая копия
# Мой код

x = np.arange(6)
part = x[1:4]
independent = x[1:4].copy()

part[0] = -10
independent[-1] = 99

print("x:", x)
print("part:", part)
print("independent:", independent)
````

> **Почему изменение `part` влияет на `x`?**
> 
> В NumPy базовая срезка (срез) возвращает **представление (view)**, а не копию данных. Переменная `part` — это просто «окно», которое смотрит на ту же самую область памяти, что и исходный массив `x`. Поэтому, когда мы меняем `part[0]`, мы на самом деле меняем элемент в памяти, на который ссылается и `x`. 
> Чтобы получить независимый массив, мы явно вызвали метод `.copy()`, который физически скопировал данные в новую область памяти — поэтому изменения в `independent` не затронули `x`.

### ★ 1.S3 Таблица умножения

Создайте таблицу умножения чисел от 1 до 10 без циклов. Подсказка: умножение массивов форм `(10, 1)` и `(1, 10)` даёт результат формы `(10, 10)`. Этот механизм называется broadcasting. Объясните, какие значения перемножаются для элемента на пересечении третьей строки и седьмого столбца.

### Задание 2.04 Строка по метке и по позиции

Из `mini` получите строку с меткой 103 и строку с позицией 1. Затем получите строки через `mini.loc[101:103]` и `mini.iloc[0:2]`. Объясните, почему границы этих срезов записаны по-разному.

````python
# Задание 2.04 Строка по метке и по позиции
# Мой код

print("Строка с меткой 103 (loc):")
print(mini.loc[103])

print("\nСтрока с позицией 1 (iloc):")
print(mini.iloc[1])

print("\nСрез mini.loc[101:103]:")
print(mini.loc[101:103])

print("\nСрез mini.iloc[0:2]:")
print(mini.iloc[0:2])
````

> **Мой ответ:**
> 
> Границы срезов записаны по-разному, потому что `loc` работает с **метками индекса**, а `iloc` — с **позициями** (номерами строк, начиная с нуля). У `loc` правая граница **включается** в результат: `mini.loc[101:103]` вернёт строки с метками 101 и 103. У `iloc` правая граница **исключается**, как в обычных срезах Python: `mini.iloc[0:2]` вернёт позиции 0 и 1 (то есть те же строки 101 и 103). Таким образом, оба среза дают одинаковый результат, но по разным правилам. Если бы индекс был не `[101, 103, 107]`, а, например, `[10, 20, 30]`, то `loc[10:20]` включил бы обе строки, а `iloc[0:1]` вернул бы только первую — тогда результаты различались бы. Различие в поведении правой границы — главное, что нужно запомнить: `loc` — «включая конец», `iloc` — «не включая конец».

### ★ 2.S2 Сложение Series по меткам

Создайте `left = pd.Series([10, 20], index=["A", "B"])` и `right = pd.Series([1, 2], index=["B", "C"])`. Предскажите и проверьте результат `left + right`. Объясните положение значений и появление пропусков. Дополнительно вычислите `left.add(right, fill_value=0)` и объясните изменение.

> **Мой ответ - предсказание результата:**
> 
> * У `left` есть метки `A` и `B`.
> * У `right` есть метки `B` и `C`.
> * Метка `B` есть в обоих Series: `20 + 1 = 21`.
> * Метки `A` (только в `left`) и `C` (только в `right`) не имеют пары, поэтому на их месте появится `NaN`.

````python
# ★ 2.S2 Сложение Series по меткам
# Мой код
left = pd.Series([10, 20], index=["A", "B"])
right = pd.Series([1, 2], index=["B", "C"])

print("Обычное сложение (left + right):")
print(left + right)

print("\nСложение с заполнением (left.add(right, fill_value=0)):")
print(left.add(right, fill_value=0))
````

> **Мой ответ - объяснения:**
> При обычном сложении `left + right` pandas выравнивает данные по меткам индексов. Метка `B` есть в обоих Series, поэтому значения складываются: 20 + 1 = 21. Метки `A` и `C` присутствуют только в одном из Series. Поскольку pandas не знает, чему равно отсутствующее значение (в `right` для `A` и в `left` для `C`), по умолчанию она подставляет `NaN` (пустое значение). Поэтому в результате для меток `A` и `C` мы видим `NaN`.
> При использовании метода `.add(fill_value=0)` мы указываем pandas: «Если метки нет во втором Series, считай, что там лежит 0». Поэтому для `A` вычисляется 10 + 0 = 10, а для `C`: 0 + 2 = 2. `NaN` пропадают, и все значения становятся числами.

### ★ 2.S3 План пополнения

Для действующих товаров рассчитайте `to_order = max(reorderLevel - unitsInStock, 0)` и стоимость такого пополнения по каталожной цене. Заказы, уже указанные в `unitsOnOrder`, в этой задаче не учитывайте. Выведите товары с положительным `to_order`, суммарное количество к заказу и общую стоимость. Подсказка: `np.maximum(series, 0)` сравнивает каждый элемент с нулём.

````python
# ★ 2.S3 План пополнения
# Мой код
# 1. Создаем копию, чтобы не менять основную таблицу
restock = products[products['discontinued'] == 0].copy()

# 2. Считаем количество к заказу (используем np.maximum)
# Формула: если (порог - остаток) > 0, то берем разницу, иначе 0
restock['to_order'] = np.maximum(restock['reorderLevel'] - restock['unitsInStock'], 0)

# 3. Считаем стоимость пополнения
restock['restock_cost'] = restock['to_order'] * restock['unitPrice']

# 4. Отбираем только те товары, которые нужно заказывать (to_order > 0)
need_to_order = restock[restock['to_order'] > 0]

print("Товары для пополнения:")
print(need_to_order[['productID', 'productName', 'unitsInStock', 'reorderLevel', 'to_order', 'restock_cost']])

print(f"\nСуммарное количество единиц к заказу: {need_to_order['to_order'].sum()}")
print(f"Общая стоимость пополнения: {need_to_order['restock_cost'].sum():.2f}")
````

````python
# ★ 3.S2 Лидеры внутри категорий

# Мой код

# Расчеты выручки 
order_details['gross_amount'] = order_details['unitPrice'] * order_details['quantity']
order_details['discount_amount'] = order_details['gross_amount'] * order_details['discount']
order_details['revenue'] = order_details['gross_amount'] - order_details['discount_amount']

# 1. Собираем выручку по товарам и категориям
product_revenue = order_details.groupby('productID', as_index=False)['revenue'].sum()

# Присоединяем категорию
product_revenue = product_revenue.merge(
    products[['productID', 'categoryID', 'productName']], 
    on='productID', how='left'
)

# 2. Вычисляем максимум выручки в каждой категории через transform
product_revenue['cat_max_revenue'] = product_revenue.groupby('categoryID')['revenue'].transform('max')

# 3. Отбираем лидеров (сравниваем через np.isclose)
mask = np.isclose(product_revenue['revenue'], product_revenue['cat_max_revenue'], rtol=0, atol=1e-6)
leaders = product_revenue[mask]

print("Товары-лидеры по выручке в своих категориях:")
print(leaders[['categoryID', 'productID', 'productName', 'revenue']].sort_values(by='categoryID'))
````

Создайте отдельную копию `orders`. Преобразуйте `orderDate` и `shippedDate` в даты через `pd.to_datetime`. Для заказов с известной датой отправки вычислите число дней между заказом и отправкой. Найдите средний срок отправки по странам и выведите пять стран с наибольшим средним. При равенстве сортируйте страны по алфавиту. Заказы без даты отправки исключите только из этого расчёта; исходную таблицу сохраните.

Подсказка: `(table['end'] - table['start']).dt.days`. Средний срок отправки считайте по заказам, до объединения с позициями.

````python
# ★ 3.S3 Срок отправки заказа
# Мой код
# 1. Создаем копию
orders_dates = orders.copy()

# 2. Преобразуем даты
orders_dates['orderDate'] = pd.to_datetime(orders_dates['orderDate'])
orders_dates['shippedDate'] = pd.to_datetime(orders_dates['shippedDate'])

# 3. Исключаем заказы без даты отправки для расчета
valid_orders = orders_dates[orders_dates['shippedDate'].notna()].copy()

# 4. Считаем дни в пути
valid_orders['shipping_days'] = (valid_orders['shippedDate'] - valid_orders['orderDate']).dt.days

# 5. Средний срок по странам
avg_shipping = valid_orders.groupby('shipCountry')['shipping_days'].mean().reset_index()

# 6. Сортируем: по убыванию среднего срока, при равенстве — по алфавиту (ascending=True)
top_slowest = avg_shipping.sort_values(
    by=['shipping_days', 'shipCountry'], 
    ascending=[False, True]
).head(5)

print("Топ-5 стран с наибольшим средним сроком отправки (в днях):")
print(top_slowest)
````
